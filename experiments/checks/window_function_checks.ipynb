```python
from swimpde_jax.domain.visualizer import visualize_window_functions
from swimpde_jax.domain import MultiLevelDecomposedDomain
import numpy as np
```

```python
SPAN = 0, 10
N_DIM = 1      # dimensions x1, x2
N_POINTS = 250
NUM_PARTITION = 4
OVERLAP_RATIO = 1.9
WINDOW_FN = "cosine" #trapezoidal, cosine, gaussian, wendland

# Sampling the points within the problem domains
points = np.linspace(*SPAN, N_POINTS).reshape(-1, N_DIM)

# Boundary
boundary_mask = (
    (points[:, 0] == SPAN[0]) |
    (points[:, 0] == SPAN[1])
)
        
# Instance definitions
domain = MultiLevelDecomposedDomain(
    interior_points=points, #points[~boundary_mask],
    boundary_points=None, #points[boundary_mask], 
    num_partition_per_level=[4],
    overlap_ratio=OVERLAP_RATIO, 
    window_fn=WINDOW_FN
)
```

```python
visualize_window_functions(domain, normalized=False, show=True, legend=True)
    
```
